Import pandas

```python
import pandas as pd
```

Load the datasets

```python
weather = pd.read_csv("/Users/eng/Documents/UIC/cs424/CS424_Project/Ye-Ng-CS424/Data/Updated_Weather_Data.csv")
study = pd.read_csv("/Users/eng/Documents/UIC/cs424/CS424_Project/Ye-Ng-CS424/Data/Updated_Study_Spots_Data.csv")
```

Merge weather and study data

```python
study["date"] = pd.to_datetime(study["date"])
weather["date"] = pd.to_datetime(weather["date"])

df = study.merge(weather, on=["date", "time"], how="left")

df.head(12)
```

Create derived variables

```python
df["capacity"] = df["occupancy"] + df["open_seats"]

df["occupancy_rate"] = (df["occupancy"] / df["capacity"]).round(3)

df["availability_rate"] = (df["open_seats"] / df["capacity"]).round(3)

df["day_of_week"] = df["date"].dt.day_name()

high_traffic = ["1st Floor Sandella Cafe", "1st Floor Lobby Area"]
df["traffic_level"] = df["location"].apply(
    lambda x: "High Traffic"
    if x in high_traffic
    else "Low Traffic"
)

df[[
    "location",
    "date",
    "occupancy",
    "open_seats",
    "capacity",
    "occupancy_rate",
    "availability_rate",
    "traffic_level"
]].head(6)
```

Check the summary statistics

```python
df.groupby("location")["capacity"].describe()
```

Filter the times

```python
times_to_keep = [
    "9:00",
    "11:00",
    "13:00",
    "15:00",
    "17:00"
]

df = df[df["time"].isin(times_to_keep)]

print(df["time"].unique())
```

Save the transformed data

```python
df.to_csv("../Data/transformed_study_data.csv", index=False)
```
